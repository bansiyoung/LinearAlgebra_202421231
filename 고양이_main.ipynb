## 고양이 그리기 — main

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
from my_utils import translate, scale, rotate, shear_x, shear_y, reflect_x, reflect_y, reflect_origin, reflect_yx

INF = 1000
LEN = 100

def draw_axis(ax):
  ax.axhline(0, color='black', alpha=0.3)  # x-axis line
  ax.axvline(0, color='black', alpha=0.3)  # y-axis line
  ax.set_xlim(-INF, INF)
  ax.set_ylim(-INF, INF)
  ax.set_aspect('equal')

def draw_rectangle(ax, p1, p2, p3, p4,color = "blue"):
  my_polygon = Polygon([p1, p2, p3, p4],color = color)
  ax.add_patch(my_polygon)

def draw_triangle(ax, p1, p2, p3, color="blue"):
    my_polygon = Polygon([p1, p2, p3], color=color)
    ax.add_patch(my_polygon)

def draw_cat_face(ax, p1, p2, p3, p4):
    # 사각형 중심을 원점으로 이동
    q1 = translate(p1, -LEN/2, -LEN/2)
    q2 = translate(p2, -LEN/2, -LEN/2)
    q3 = translate(p3, -LEN/2, -LEN/2)
    q4 = translate(p4, -LEN/2, -LEN/2)

    # 크기 변환: 얼굴을 가로 400, 세로 300으로 확대
    q1 = scale(q1, 4, 3)
    q2 = scale(q2, 4, 3)
    q3 = scale(q3, 4, 3)
    q4 = scale(q4, 4, 3)

    draw_rectangle(ax, q1, q2, q3, q4, color="orange")

def draw_cat_ears(ax):
    # 왼쪽 귀
    p1 = np.array([-200, 150])
    p2 = np.array([-180, 400])
    p3 = np.array([-20, 150])

    draw_triangle(ax, p1, p2, p3, color="orange")

    # y축 반사변환: 오른쪽 귀 생성
    q1 = reflect_y(p1)
    q2 = reflect_y(p2)
    q3 = reflect_y(p3)

    draw_triangle(ax, q1, q2, q3, color="orange")

    # 왼쪽 귀 안쪽
    r1 = np.array([-160, 170])
    r2 = np.array([-165, 320])
    r3 = np.array([-60, 170])

    draw_triangle(ax, r1, r2, r3, color="pink")

    # 오른쪽 귀 안쪽
    draw_triangle(
        ax, reflect_y(r1), reflect_y(r2), reflect_y(r3),
        color="pink"
    )

def draw_cat_eyes(ax, p1, p2, p3, p4):
    # 중심을 원점으로 이동
    q1 = translate(p1, -LEN/2, -LEN/2)
    q2 = translate(p2, -LEN/2, -LEN/2)
    q3 = translate(p3, -LEN/2, -LEN/2)
    q4 = translate(p4, -LEN/2, -LEN/2)

    # 크기 변환: 눈을 작게 만듦
    q1 = scale(q1, 0.35, 0.6)
    q2 = scale(q2, 0.35, 0.6)
    q3 = scale(q3, 0.35, 0.6)
    q4 = scale(q4, 0.35, 0.6)

    # 이동 변환: 왼쪽 눈 배치
    q1 = translate(q1, -90, 40)
    q2 = translate(q2, -90, 40)
    q3 = translate(q3, -90, 40)
    q4 = translate(q4, -90, 40)

    draw_rectangle(ax, q1, q2, q3, q4, color="black")

    # y축 반사변환: 오른쪽 눈 생성
    r1 = reflect_y(q1)
    r2 = reflect_y(q2)
    r3 = reflect_y(q3)
    r4 = reflect_y(q4)

    draw_rectangle(ax, r1, r2, r3, r4, color="black")

def draw_cat_nose(ax):
    p1 = np.array([-35, -30])
    p2 = np.array([35, -30])
    p3 = np.array([0, -75])

    draw_triangle(ax, p1, p2, p3, color="pink")

def draw_cat_whiskers(ax):
    # 입
    ax.plot([0, 0], [-75, -105], color="black", linewidth=3)
    ax.plot([-45, 0, 45], [-110, -130, -110],
            color="black", linewidth=3)

    # 왼쪽 수염의 시작점과 끝점
    whiskers = [
        ([-110, -280], [-55, -10]),
        ([-110, -290], [-85, -85]),
        ([-110, -280], [-115, -165])
    ]

    for xs, ys in whiskers:
        ax.plot(xs, ys, color="black", linewidth=3)

        # y축 반사변환으로 오른쪽 수염 생성
        p1 = reflect_y(np.array([xs[0], ys[0]]))
        p2 = reflect_y(np.array([xs[1], ys[1]]))

        ax.plot([p1[0], p2[0]],
                [p1[1], p2[1]],
                color="black", linewidth=3)

def draw_object(ax):
    p1 = np.array([0, 0])
    p2 = np.array([LEN, 0])
    p3 = np.array([LEN, LEN])
    p4 = np.array([0, LEN])

    draw_cat_ears(ax)
    draw_cat_face(ax, p1, p2, p3, p4)
    draw_cat_eyes(ax, p1, p2, p3, p4)
    draw_cat_nose(ax)
    draw_cat_whiskers(ax)

# main: 그림판 생성 및 고양이 그리기
fig, ax = plt.subplots(figsize=(8, 8))
draw_object(ax)
ax.set_xlim(-500, 500)
ax.set_ylim(-500, 500)
ax.set_aspect("equal")
ax.grid(True)
plt.show()
```
